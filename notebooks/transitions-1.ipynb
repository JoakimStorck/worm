```python
import pandas as pd
import numpy as np

# Filvägar (justera om de ligger på annan plats)
path_labels = "delRioChanona/ipums_labs_mSML_manual.csv"
path_matrix = "delRioChanona/occupational_mobility_network.csv"

# Läs in etiketter (yrkesnamn i rätt ordning)
labels_df = pd.read_csv(path_labels)
labels = labels_df['label'].tolist()

# Läs in sannolikhetsmatris
P_raw = pd.read_csv(path_matrix, header=None)
assert len(labels) == P_raw.shape[0] == P_raw.shape[1], "Antal yrken matchar inte matrisens dimensioner"

# Skapa namngiven DataFrame
P = pd.DataFrame(P_raw.values, index=labels, columns=labels)

# Exempel: visa sannolikhet att gå från bartender till chef
print("P(bartender → chef):", P.loc["bartenders", "chefs and head cooks"])

# Visa de fem yrken med högst sannolikhet att lämna ett visst yrke
def top_transitions(from_occupation, n=5):
    return P.loc[from_occupation].sort_values(ascending=False).head(n)

# Exempel
top_transitions("bartenders")
```

```python
import matplotlib.pyplot as plt
import networkx as nx

threshold = 0.01  # visa bara länkar med P_ij > 1%
G = nx.DiGraph()

for i, from_occ in enumerate(P.index):
    for j, to_occ in enumerate(P.columns):
        prob = P.iloc[i, j]
        if prob > threshold:
            G.add_edge(from_occ, to_occ, weight=prob)

plt.figure(figsize=(12, 12))
pos = nx.spring_layout(G, k=0.15, iterations=50)
nx.draw(G, pos, node_size=10, edge_color='gray', alpha=0.3, with_labels=False)
plt.title(f"Yrkesnätverk: övergångar med sannolikhet > {threshold}")
plt.show()
```
